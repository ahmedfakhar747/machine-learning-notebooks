# 🤔 What is Pandas Profiling?

Pandas Profiling (now called **YData Profiling**) is an automated Exploratory Data Analysis (EDA) library.

With a single line of code, it generates an interactive report containing:

- 📊 Dataset overview
- 📈 Summary statistics
- ❓ Missing values
- 🔄 Duplicate rows
- 🔥 Correlations
- 📉 Distributions
- ⚠️ Outliers
- 🔤 Data types
- 🎯 Variable insights

It saves hours of manual analysis.
