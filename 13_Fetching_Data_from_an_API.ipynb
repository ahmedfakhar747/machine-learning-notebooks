# 💻 ML Lab 04 — Fetching Data from an API

```python
import pandas as pd
import requests
```

```python
response = requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page=1")
```

```python
columns = [
    "id",
    "title",
    "original_title",
    "original_language",
    "genre_ids",
    "overview",
    "release_date",
    "popularity",
    "vote_average",
    "vote_count"
]
```

```python
df = pd.DataFrame(response.json()["results"])[columns]
df.head()
```

```python
df1 = pd.DataFrame()
```

```python
for i in range(1, 500):
    response = requests.get(f'https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={i}')
    temp_df = pd.DataFrame(response.json()['results'])[columns]
    df1 = pd.concat([df1, temp_df] ,ignore_index=True)
```

```python
df1
```

```python
df1.to_csv("movies_metadata.csv")
```
