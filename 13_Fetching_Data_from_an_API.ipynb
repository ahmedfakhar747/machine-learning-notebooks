# 📦 Import Libraries



````python
import pandas as pd
import requests
````

---

# 🌐 Understanding the API

We'll use **The Movie Database (TMDB)** API, which provides information about thousands of movies.

The endpoint we'll use returns the **Top Rated Movies**.

```
https://api.themoviedb.org/3/movie/top_rated
```

The API returns data in **JSON** format.



---

# 🚀 Step 1 — Make Your First API Request


````python
response = requests.get(
    "https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page=1"
)
````


---

# 🔍 Step 2 — Inspect the JSON Response

Let's see what the API returns.


````python
response.json()
````


You'll notice several keys such as:

* page
* results
* total_pages
* total_results

The actual movie data is stored inside:



````python
response.json()["results"]
````


---

# 📑 Step 3 — Select Required Columns

Instead of storing everything, we'll keep only the most useful columns.



````python
columns = [
    "id",
    "title",
    "original_title",
    "original_language",
    "genre_ids",
    "overview",
    "release_date",
    "popularity",
    "vote_average",
    "vote_count"
]
````

---

# 🐼 Step 4 — Fetch Multiple Pages

Now we'll collect movies from multiple API pages.





````python
df = pd.DataFrame()

for page in range(1, 500):

    response = requests.get(
        f"https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={page}"
    )

    temp_df = pd.DataFrame(response.json()["results"])[columns]

    df = pd.concat([df, temp_df], ignore_index=True)
````

---

# 👀 Step 5 — View the Dataset


````python
df
````

---

# 💾 Step 6 — Save the Dataset


````python
df.to_csv("movies_metadata.csv", index=False)
````


Now the collected API data has been converted into a CSV file that can be used for:

* 📊 Data Analysis
* 🤖 Machine Learning
* 📈 Visualization
* 🔍 Exploratory Data Analysis (EDA)

# ⚠️ Things to Keep in Mind

* Never expose your API key publicly.
* Many APIs have **rate limits**, so avoid sending too many requests in a short time.
* Always check the API documentation before using an endpoint.
* APIs may return missing values or errors, so production code should include proper error handling.
