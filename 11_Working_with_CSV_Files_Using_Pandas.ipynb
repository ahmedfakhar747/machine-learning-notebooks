# 5️⃣ Selecting Data

Selecting specific rows and columns is one of the most common tasks when working with datasets. Pandas provides several ways to access data efficiently.

## 5.1 Selecting a Single Column


````python
df["cgpa"]
````


## 5.2 Selecting Multiple Columns


````python
df[["cgpa", "placement_exam_marks"]]
````


## 5.3 Selecting Rows using `loc`

Select rows by their labels.



````python
df.loc[0]
````

````python
df.loc[0:4]
````

````python
df.loc[:, ["cgpa", "placed"]]
````

## 5.4 Selecting Rows using `iloc`

Select rows by their integer positions.



````python
df.iloc[0]
````

````python
df.iloc[0:5]
````

````python
df.iloc[:, 0:2]
````

---

# 6️⃣ Filtering Data

Filtering allows us to retrieve only the records that satisfy specific conditions.

## 6.1 Conditional Filtering

Students with CGPA greater than 8.



````python
df[df["cgpa"] > 8]
````

## 6.2 Multiple Conditions

Students with CGPA greater than 7 and placed.



````python
df[(df["cgpa"] > 7) & (df["placed"] == 1)]
````


Students with CGPA greater than 8 or placed.



````python
df[(df["cgpa"] > 8) | (df["placed"] == 1)]
````



## 6.3 Boolean Indexing

Store the condition in a Boolean variable.





````python
high_cgpa = df["cgpa"] > 8

df[high_cgpa]
````

---

# 7️⃣ Missing Values

Missing values are common in real-world datasets. Pandas makes it easy to detect them.

## 7.1 Check Missing Values

```python
df.isnull()
```


## 7.2 Count Missing Values

```python
df.isnull().sum()
```



---

# 8️⃣ Duplicate Values

Duplicate records can negatively affect Machine Learning models, so they should be identified and removed.

## 8.1 Check Duplicate Rows

```python
df.duplicated()
```



## 8.2 Count Duplicate Rows

```python
df.duplicated().sum()
```



## 8.3 Remove Duplicate Rows

```python
df.drop_duplicates(inplace=True)
```



---

# 9️⃣ Creating New Columns

Feature engineering often starts by creating new columns from existing data.

## 9.1 Create a New Column

```python
df["total_score"] = df["cgpa"] + df["placement_exam_marks"]

df.head()
```



## 9.2 Modify an Existing Column

Increase CGPA by 0.1 (for demonstration).

```python
df["cgpa"] = df["cgpa"] + 0.1

df.head()
```



## 9.3 Delete a Column

```python
df.drop(columns=["total_score"], inplace=True)
```
