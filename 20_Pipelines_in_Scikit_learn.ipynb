# 🚀 Introduction

As machine learning projects become more complex, managing multiple preprocessing steps and model training separately can make your code repetitive and difficult to maintain.

**Pipelines** in **Scikit-learn** solve this problem by chaining multiple steps—such as **data preprocessing**, **feature transformation**, and **model training**—into a single workflow. This makes your code cleaner, reduces the risk of data leakage, and ensures that the same preprocessing is consistently applied during both training and prediction.

In this notebook, you'll learn how to build machine learning pipelines in **Scikit-learn**, understand why they're important, and use them to create efficient, organized, and production-ready ML workflows with practical examples. 🚀

# 🧩 What is a Pipeline?

A **Pipeline** is a sequence of transformers followed by a final estimator (machine learning model).



Each step passes its output to the next step.


````python
'''
Raw Data
    │
    ▼
Missing Value Imputation
    │
    ▼
Encoding
    │
    ▼
Feature Scaling
    │
    ▼
Feature Selection (Optional)
    │
    ▼
Machine Learning Model
    │
    ▼
Predictions
'''
````

# 📦 Importing Libraries

````python
import pandas as pd
import numpy as np

from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (OneHotEncoder, MinMaxScaler)
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.tree import DecisionTreeClassifier
````

````python
df = pd.read_csv("/kaggle/input/datasets/ahmedfakhar123/titanic-survival-dataset/Titanic_Survival_Dataset.csv")
df
````

````python
df.drop(columns=["PassengerId", "Name", "Ticket", "Cabin"], inplace=True)
df
````

````python
X = df.drop(columns=["Survived"])
y = df.Survived
````

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
````

---

# 🚀 Building the Machine Learning Pipeline

Instead of manually applying each preprocessing step one by one, we'll combine everything into a **single Scikit-learn Pipeline**. This ensures that every transformation is executed in the correct order during training and prediction.

Our pipeline will perform the following operations:

1. 🩹 Handle missing values
2. 🔤 Encode categorical features
3. 📏 Scale numerical features
4. 🎯 Select the most important features
5. 🌳 Train a Decision Tree classifier

This approach keeps the workflow clean, reusable, and helps prevent data leakage.




# 🩹 Step 1 — Handling Missing Values

Machine learning models cannot work directly with missing values (`NaN`). Before training, we need to replace them with meaningful values.

In this dataset:

* **Age** is filled using the **mean** value.
* **Embarked** is filled using the **most frequent** category.




````python
# Step 1: Imputation
trf1 = ColumnTransformer([
    ('impute_age', SimpleImputer(), [2]),
    ('impute_embarked', SimpleImputer(strategy='most_frequent'), [6])
], remainder='passthrough')
````


# 🔤 Step 2 — Encoding Categorical Features

Machine learning algorithms require numerical input, but our dataset contains categorical features such as **Sex** and **Embarked**.

We'll use **One-Hot Encoding** to convert each category into separate binary columns while avoiding issues with unseen categories during prediction.

````python
# Step 2: Encoding
trf2 = ColumnTransformer([
    ('encode_sex_embarked', OneHotEncoder(sparse_output=False, handle_unknown='ignore'), [1, 3])
], remainder='passthrough')
````

# 📏 Step 3 — Feature Scaling

Features often have very different ranges.

For example:

* Age → around **20–80**
* Fare → can exceed **500**

We'll normalize the features using **MinMaxScaler**, which rescales values to the range **0–1**.

````python
# Step 3: Scaling
trf3 = ColumnTransformer([
    ('scale', MinMaxScaler(), slice(0, 10)),
], remainder='passthrough')
````


# 🎯 Step 4 — Feature Selection

Not every feature contributes equally to predicting survival.

Using **SelectKBest** with the **Chi-Square (χ²)** statistical test, we'll automatically select the most informative features before training the model.

````python
# Step 4: Feature Selection
trf4 = SelectKBest(score_func=chi2, k=8)
````

# 🌳 Step 5 — Machine Learning Model

Finally, we'll train a **Decision Tree Classifier** using the transformed features.



````python
# Step 5: Machine Learning Model
trf5 = DecisionTreeClassifier()
````


# 🔗 Combining Everything into a Pipeline

Now we'll combine all preprocessing steps and the machine learning model into a single Pipeline.

From this point onward, calling **`fit()`** or **`predict()`** on the pipeline automatically executes every step in the correct order.



````python
pipe = Pipeline([
    ('trf1', trf1),
    ('trf2', trf2),
    ('trf3', trf3),
    ('trf4', trf4),
    ('trf5', trf5)
])
````

# ⚡ Pipeline vs `make_pipeline()`

Scikit-learn provides two ways to create a pipeline.

### ✅ `Pipeline`

With `Pipeline`, you explicitly assign a name to every step.

```python
pipe = Pipeline([
    ("preprocessing", preprocessor),
    ("model", DecisionTreeClassifier())
])
```

**Advantages**

* Custom step names
* Easier debugging
* Required when referencing specific steps (e.g., GridSearchCV parameters)


### ✅ `make_pipeline()`

`make_pipeline()` automatically generates step names based on the estimator class names.



````python
pipe = make_pipeline(trf1, trf2, trf3, trf4, trf5)
````


**Advantages**

* Less code
* Faster to write
* Great for quick experiments


### 📌 Which one should you use?

For educational notebooks and production projects, **`Pipeline` is generally preferred** because the custom step names make the workflow easier to understand and simplify hyperparameter tuning with `GridSearchCV`.

# 🚀 Training the Pipeline

Training the pipeline is as simple as training any other Scikit-learn model.

A single call to `fit()` performs every preprocessing step automatically before fitting the Decision Tree classifier.



````python
pipe.fit(X_train, y_train)
````

# 🔍 Exploring the Pipeline

After training, you can inspect individual components using the `named_steps` attribute. This is useful for understanding the pipeline, debugging preprocessing steps, or accessing learned parameters.



````python
pipe.named_steps
````

For example, you can inspect the first `ColumnTransformer`:



````python
pipe.named_steps['columntransformer-1']
````

Or retrieve the most frequent value learned by the `SimpleImputer`:

````python
pipe.named_steps['columntransformer-1'].transformers_[1][1].statistics_
````

This demonstrates that a pipeline is more than a black box—you can still access and inspect each preprocessing step whenever needed.

Predicting

````python
y_pred = pipe.predict(X_test)
y_pred
````

````python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
````

Cross Validation using Pipeline

````python
from sklearn.model_selection import cross_val_score
cross_val_score(pipe, X_train, y_train, cv=5, scoring='accuracy').mean()
````

GridSearch using Pipeline

````python
# gridsearchcv
params = {
    'decisiontreeclassifier__max_depth': [1, 2, 3, 4, 5, None]
}

from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(pipe, params, cv=5, scoring='accuracy')
grid.fit(X_train, y_train)
````

````python
grid.best_score_
````

````python
grid.best_params_
````
