# 🚀 Introduction

As machine learning projects become more complex, managing multiple preprocessing steps and model training separately can make your code repetitive and difficult to maintain.

**Pipelines** in **Scikit-learn** solve this problem by chaining multiple steps—such as **data preprocessing**, **feature transformation**, and **model training**—into a single workflow. This makes your code cleaner, reduces the risk of data leakage, and ensures that the same preprocessing is consistently applied during both training and prediction.

In this notebook, you'll learn how to build machine learning pipelines in **Scikit-learn**, understand why they're important, and use them to create efficient, organized, and production-ready ML workflows with practical examples. 🚀

# 🎯 Why Use Pipelines?

Without pipelines, your workflow might look like this:



````python
# Handle missing values
# Encode categorical data
# Scale features
# Train model
# Make predictions
````

As your project grows, this quickly becomes messy.

With a pipeline:

````python
# pipeline.fit(X_train, y_train)
# predictions = pipeline.predict(X_test)
````

Everything happens automatically in the correct order.

---
# 🧩 What is a Pipeline?

A **Pipeline** is a sequence of transformers followed by a final estimator (machine learning model).



Each step passes its output to the next step.


````python
'''
Raw Data
    │
    ▼
Missing Value Imputation
    │
    ▼
Encoding
    │
    ▼
Feature Scaling
    │
    ▼
Feature Selection (Optional)
    │
    ▼
Machine Learning Model
    │
    ▼
Predictions
'''
````


---

# ✅ Advantages of Pipelines

* 🚀 Cleaner and more organized code
* 🔒 Prevents data leakage
* 🔁 Reproducible workflows
* 🛠 Easy to modify preprocessing steps
* 🎯 Works seamlessly with Cross Validation
* 🔍 Compatible with GridSearchCV and RandomizedSearchCV
* 📦 Easy to save and deploy

# 📦 Importing Libraries

````python
import pandas as pd
import numpy as np

from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (OneHotEncoder, MinMaxScaler)
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.tree import DecisionTreeClassifier
````

---
# 📂 Loading the Dataset

First, let's load the Titanic Survival Dataset into a Pandas DataFrame.

````python
df = pd.read_csv("/kaggle/input/datasets/ahmedfakhar123/titanic-survival-dataset/Titanic_Survival_Dataset.csv")
df
````

---
# 🧹 Removing Unnecessary Features

Some columns are either unique identifiers, text-based information, or contain many missing values. Since they aren't needed for this notebook, we'll remove them to simplify our preprocessing pipeline.

````python
df.drop(columns=["PassengerId", "Name", "Ticket", "Cabin"], inplace=True)
df
````

This leaves us with the features required to build our machine learning pipeline.

---

# ✂️ Splitting Features and Target

Before training our model, we need to separate the **input features** from the **target variable**.

* **`X`** contains all the input features used to make predictions.
* **`y`** contains the target variable (**`Survived`**), which the model will learn to predict.

Next, we'll split the data into **training** and **testing** sets. The model is trained on the training set and evaluated on the unseen testing set to measure how well it generalizes to new data.

````python
X = df.drop(columns=["Survived"])
y = df.Survived
````

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
````

Now our data is ready for preprocessing and building the machine learning pipeline.

---

# 🚀 Building the Machine Learning Pipeline

Instead of manually applying each preprocessing step one by one, we'll combine everything into a **single Scikit-learn Pipeline**. This ensures that every transformation is executed in the correct order during training and prediction.

Our pipeline will perform the following operations:

1. 🩹 Handle missing values
2. 🔤 Encode categorical features
3. 📏 Scale numerical features
4. 🎯 Select the most important features
5. 🌳 Train a Decision Tree classifier

This approach keeps the workflow clean, reusable, and helps prevent data leakage.




# 🩹 Step 1 — Handling Missing Values

Machine learning models cannot work directly with missing values (`NaN`). Before training, we need to replace them with meaningful values.

In this dataset:

* **Age** is filled using the **mean** value.
* **Embarked** is filled using the **most frequent** category.




````python
# Step 1: Imputation
trf1 = ColumnTransformer([
    ('impute_age', SimpleImputer(), [2]),
    ('impute_embarked', SimpleImputer(strategy='most_frequent'), [6])
], remainder='passthrough')
````


# 🔤 Step 2 — Encoding Categorical Features

Machine learning algorithms require numerical input, but our dataset contains categorical features such as **Sex** and **Embarked**.

We'll use **One-Hot Encoding** to convert each category into separate binary columns while avoiding issues with unseen categories during prediction.

````python
# Step 2: Encoding
trf2 = ColumnTransformer([
    ('encode_sex_embarked', OneHotEncoder(sparse_output=False, handle_unknown='ignore'), [1, 3])
], remainder='passthrough')
````

# 📏 Step 3 — Feature Scaling

Features often have very different ranges.

For example:

* Age → around **20–80**
* Fare → can exceed **500**

We'll normalize the features using **MinMaxScaler**, which rescales values to the range **0–1**.

````python
# Step 3: Scaling
trf3 = ColumnTransformer([
    ('scale', MinMaxScaler(), slice(0, 10)),
], remainder='passthrough')
````


# 🎯 Step 4 — Feature Selection

Not every feature contributes equally to predicting survival.

Using **SelectKBest** with the **Chi-Square (χ²)** statistical test, we'll automatically select the most informative features before training the model.

````python
# Step 4: Feature Selection
trf4 = SelectKBest(score_func=chi2, k=8)
````

# 🌳 Step 5 — Machine Learning Model

Finally, we'll train a **Decision Tree Classifier** using the transformed features.



````python
# Step 5: Machine Learning Model
trf5 = DecisionTreeClassifier()
````


# 🔗 Combining Everything into a Pipeline

Now we'll combine all preprocessing steps and the machine learning model into a single Pipeline.

From this point onward, calling **`fit()`** or **`predict()`** on the pipeline automatically executes every step in the correct order.



````python
pipe = Pipeline([
    ('trf1', trf1),
    ('trf2', trf2),
    ('trf3', trf3),
    ('trf4', trf4),
    ('trf5', trf5)
])
````

# ⚡ Pipeline vs `make_pipeline()`

Scikit-learn provides two ways to create a pipeline.

### ✅ `Pipeline`

With `Pipeline`, you explicitly assign a name to every step.

```python
pipe = Pipeline([
    ("preprocessing", preprocessor),
    ("model", DecisionTreeClassifier())
])
```

**Advantages**

* Custom step names
* Easier debugging
* Required when referencing specific steps (e.g., GridSearchCV parameters)


### ✅ `make_pipeline()`

`make_pipeline()` automatically generates step names based on the estimator class names.



````python
pipe = make_pipeline(trf1, trf2, trf3, trf4, trf5)
````


**Advantages**

* Less code
* Faster to write
* Great for quick experiments


### 📌 Which one should you use?

For educational notebooks and production projects, **`Pipeline` is generally preferred** because the custom step names make the workflow easier to understand and simplify hyperparameter tuning with `GridSearchCV`.

Scikit-learn can display pipelines as interactive diagrams, making it easier to understand how data flows through each preprocessing step before reaching the final model.



````python
# Display Pipeline

from sklearn import set_config
set_config(display='diagram')
````

````python
pipe
````

---
# 🚀 Training the Pipeline

Training the pipeline is as simple as training any other Scikit-learn model.

A single call to `fit()` performs every preprocessing step automatically before fitting the Decision Tree classifier.



````python
pipe.fit(X_train, y_train)
````

---
# 🔍 Exploring the Pipeline

After training, you can inspect individual components using the `named_steps` attribute. This is useful for understanding the pipeline, debugging preprocessing steps, or accessing learned parameters.



````python
pipe.named_steps
````

For example, you can inspect the first `ColumnTransformer`:



````python
pipe.named_steps['columntransformer-1']
````

Or retrieve the most frequent value learned by the `SimpleImputer`:

````python
pipe.named_steps['columntransformer-1'].transformers_[1][1].statistics_
````

This demonstrates that a pipeline is more than a black box—you can still access and inspect each preprocessing step whenever needed.

---
# 🔮 Making Predictions

Now that our pipeline has been trained, we can use it to predict the survival status of passengers in the test dataset.

The pipeline automatically performs all preprocessing steps before making predictions, so there's no need to manually impute missing values, encode categorical features, or scale the data.

````python
y_pred = pipe.predict(X_test)
y_pred
````

---
# 📊 Evaluating Model Performance

To measure how well our model performs, we'll calculate the accuracy score, which represents the percentage of correctly classified samples.

````python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
````

The model achieves an accuracy of approximately **83%**, indicating that it correctly predicts the survival outcome for most passengers in the test set

---
# 🔄 Cross Validation with Pipeline

A single train-test split may not always provide a reliable estimate of model performance. Cross-validation evaluates the pipeline on multiple data splits, giving a more robust measure of its generalization ability.

Since all preprocessing steps are inside the pipeline, they are applied independently within each fold, preventing **data leakage**.

````python
from sklearn.model_selection import cross_val_score
cross_val_score(pipe, X_train, y_train, cv=5, scoring='accuracy').mean()
````

---
# 🎯 Hyperparameter Tuning with GridSearchCV

Instead of manually trying different hyperparameter values, we can use **GridSearchCV** to automatically search for the best combination.

Here, we'll tune the **maximum depth** of the Decision Tree classifier.

````python
# gridsearchcv
params = {
    'decisiontreeclassifier__max_depth': [1, 2, 3, 4, 5, None]
}

from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(pipe, params, cv=5, scoring='accuracy')
grid.fit(X_train, y_train)
````

After training, let's view the best cross-validation score and the optimal parameter.

````python
grid.best_score_
````

````python
grid.best_params_
````

The double underscore (`__`) is used to access parameters of a specific step inside the pipeline

---
# 💾 Saving the Pipeline

Once the pipeline is trained, we can save it to disk using **Pickle**. This allows us to reuse the entire preprocessing workflow and trained model without retraining.

````python
import pickle

pickle.dump(pipe, open("pipe.pkl", "wb"))
````

---
# 📂 Loading the Saved Pipeline

The saved pipeline can be loaded whenever needed.

````python
loaded_pipe = pickle.load(open("pipe.pkl", "rb"))
````

---
# 🚀 Predicting on New Data

Finally, let's use the loaded pipeline to predict the survival of a new passenger.

Notice that we only provide the **raw input features**. The pipeline automatically performs all preprocessing steps before making the prediction.

````python
test_input = pd.DataFrame({
    "Pclass": [1],
    "Sex": ["female"],
    "Age": [21],
    "SibSp": [1],
    "Parch": [2],
    "Fare": [50],
    "Embarked": ["S"]
})
````

````python
loaded_pipe.predict(test_input)
````

This demonstrates one of the biggest advantages of Scikit-learn Pipelines: **the complete preprocessing workflow and trained model are packaged into a single reusable object**, making prediction and deployment much simpler. 🚀
