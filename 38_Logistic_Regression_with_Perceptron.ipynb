<div style="border-bottom: 2px solid #00d4ff; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #00d4ff; color: #0f2027; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 1</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">What is Logistic Regression?</h2>
</div>

Despite the name, **Logistic Regression is a classification algorithm**, not a regression one. Instead of predicting a continuous value, it predicts the **probability** that a sample belongs to a class — then thresholds that probability to make a binary decision.

<div style="background: #0d1f2d; border: 1px solid #1e4d6b; border-radius: 12px; padding: 24px 32px; margin: 20px 0;">
  <div style="color: #7ecfdf; font-size: 11px; letter-spacing: 1.5px; text-transform: uppercase; margin-bottom: 14px;">Core Idea — Two Steps</div>
  <div style="display: grid; grid-template-columns: 1fr 1fr; gap: 20px;">
    <div>
      <div style="color: #00d4ff; font-weight: 700; font-size: 13px; margin-bottom: 6px;">① Linear combination</div>
      <div style="color: #e0f0f8; font-size: 16px; font-family: Georgia, serif;">z = θ₀ + θ₁x₁ + θ₂x₂ + ...</div>
      <div style="color: #7ecfdf; font-size: 12px; margin-top: 6px;">Same as linear regression — a weighted sum of inputs</div>
    </div>
    <div>
      <div style="color: #00d4ff; font-weight: 700; font-size: 13px; margin-bottom: 6px;">② Sigmoid squash → probability</div>
      <div style="color: #e0f0f8; font-size: 16px; font-family: Georgia, serif;">σ(z) = 1 / (1 + e⁻ᶻ)</div>
      <div style="color: #7ecfdf; font-size: 12px; margin-top: 6px;">Maps any real number to (0, 1) — a probability</div>
    </div>
  </div>
  <div style="margin-top: 16px; padding-top: 14px; border-top: 1px solid #1e4d6b; color: #00d4ff; font-size: 15px; font-family: Georgia, serif; text-align: center;">
    ŷ = 1 if σ(z) ≥ 0.5 &nbsp;&nbsp;|&nbsp;&nbsp; ŷ = 0 if σ(z) &lt; 0.5
  </div>
</div>

<div style="display: flex; gap: 14px; margin: 16px 0; flex-wrap: wrap;">
  <div style="flex: 1; min-width: 180px; background: #0d1f2d; border: 1px solid #1e4d6b; border-radius: 8px; padding: 14px 16px;">
    <div style="color: #00d4ff; font-weight: 700; font-size: 13px; margin-bottom: 4px;">📊 Output</div>
    <div style="color: #7ecfdf; font-size: 13px;">A probability between 0 and 1, not a raw class label</div>
  </div>
  <div style="flex: 1; min-width: 180px; background: #0d1f2d; border: 1px solid #1e4d6b; border-radius: 8px; padding: 14px 16px;">
    <div style="color: #00d4ff; font-weight: 700; font-size: 13px; margin-bottom: 4px;">📐 Boundary</div>
    <div style="color: #7ecfdf; font-size: 13px;">Creates a linear decision boundary in feature space</div>
  </div>
  <div style="flex: 1; min-width: 180px; background: #0d1f2d; border: 1px solid #1e4d6b; border-radius: 8px; padding: 14px 16px;">
    <div style="color: #00d4ff; font-weight: 700; font-size: 13px; margin-bottom: 4px;">⚙️ Training</div>
    <div style="color: #7ecfdf; font-size: 13px;">Minimises Log Loss (Binary Cross-Entropy), not MSE</div>
  </div>
</div>

<a id='perceptron-concept'></a>
<div style="border-bottom: 2px solid #00d4ff; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #00d4ff; color: #0f2027; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 2</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">What is a Perceptron?</h2>
</div>

The **Perceptron** (Frank Rosenblatt, 1958) is the oldest and simplest classification algorithm — and the direct ancestor of modern neural networks. It models a single biological neuron: receives inputs, computes a weighted sum, and fires (1) or stays silent (0) based on a threshold.

<div style="background: #0d1f2d; border: 1px solid #1e4d6b; border-radius: 12px; padding: 22px 28px; margin: 20px 0;">
  <div style="color: #7ecfdf; font-size: 11px; letter-spacing: 1.5px; text-transform: uppercase; margin-bottom: 12px;">Perceptron Learning Rule</div>
  <div style="color: #00d4ff; font-size: 18px; font-family: Georgia, serif; text-align: center; margin-bottom: 14px;">
    w ← w + η · (y − ŷ) · x
  </div>
  <div style="display: grid; grid-template-columns: 1fr 1fr 1fr; gap: 12px; font-size: 12px;">
    <div style="text-align: center;">
      <div style="color: #00d4ff; font-weight: 700;">η (eta)</div>
      <div style="color: #7ecfdf; margin-top: 3px;">Learning rate — step size</div>
    </div>
    <div style="text-align: center;">
      <div style="color: #00d4ff; font-weight: 700;">(y − ŷ)</div>
      <div style="color: #7ecfdf; margin-top: 3px;">Error — 0 if correct, ±1 if wrong</div>
    </div>
    <div style="text-align: center;">
      <div style="color: #00d4ff; font-weight: 700;">x</div>
      <div style="color: #7ecfdf; margin-top: 3px;">Input vector for that sample</div>
    </div>
  </div>
</div>

<div style="display: grid; grid-template-columns: 1fr 1fr; gap: 14px; margin: 16px 0;">
  <div style="background: #0d1f2d; border: 1px solid #1e4d6b; border-top: 2px solid #00d4ff; border-radius: 0 0 8px 8px; padding: 14px 16px;">
    <div style="color: #00d4ff; font-weight: 700; font-size: 13px; margin-bottom: 6px;">✅ Similarity to Logistic Regression</div>
    <div style="color: #7ecfdf; font-size: 13px; line-height: 1.6;">Both are linear classifiers. Both learn a weight vector that defines a hyperplane separating the two classes.</div>
  </div>
  <div style="background: #0d1f2d; border: 1px solid #1e4d6b; border-top: 2px solid #e74c3c; border-radius: 0 0 8px 8px; padding: 14px 16px;">
    <div style="color: #e74c3c; font-weight: 700; font-size: 13px; margin-bottom: 6px;">🔑 Key Difference</div>
    <div style="color: #7ecfdf; font-size: 13px; line-height: 1.6;">Perceptron uses a hard step function (0 or 1). Logistic Regression uses a soft sigmoid — giving probabilities, not just class labels.</div>
  </div>
</div>

We use `make_classification` to generate a **linearly separable** 2D binary dataset. The key parameters that make this work for a Perceptron:

<div style="background: #0d1f2d; border: 1px solid #1e4d6b; border-radius: 10px; padding: 18px 24px; margin: 16px 0; font-size: 13px;">

<a id='step'></a>
<div style="border-bottom: 2px solid #00d4ff; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #00d4ff; color: #0f2027; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 5</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">Step Activation Function</h2>
</div>

The Perceptron uses a **Heaviside step function** as its activation — it outputs 1 if the weighted sum is positive, 0 otherwise. No gradients, no probabilities — just a hard binary decision.

This is what distinguishes it from Logistic Regression, which uses a smooth sigmoid that allows gradient-based learning.

```python
def step(z):
    """Heaviside step function — Perceptron activation."""
    return 1 if z > 0 else 0

# Visualise both activation functions side-by-side
z_vals  = np.linspace(-6, 6, 300)
sigmoid = 1 / (1 + np.exp(-z_vals))
step_fn = np.array([step(z) for z in z_vals])

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(z_vals, step_fn, color='#00d4ff', linewidth=2.5)
axes[0].axvline(0, color='gray', linewidth=0.8, linestyle='--')
axes[0].axhline(0.5, color='gray', linewidth=0.8, linestyle='--')
axes[0].set_title('Step Function — Perceptron', fontsize=12, fontweight='bold')
axes[0].set_xlabel('z = w·x + b')
axes[0].set_ylabel('Output')
axes[0].set_yticks([0, 0.5, 1])
axes[0].grid(True, alpha=0.3)

axes[1].plot(z_vals, sigmoid, color='#f0a500', linewidth=2.5)
axes[1].axvline(0, color='gray', linewidth=0.8, linestyle='--')
axes[1].axhline(0.5, color='gray', linewidth=0.8, linestyle='--')
axes[1].set_title('Sigmoid Function — Logistic Regression', fontsize=12, fontweight='bold')
axes[1].set_xlabel('z = w·x + b')
axes[1].set_ylabel('σ(z) = Probability')
axes[1].set_yticks([0, 0.5, 1])
axes[1].grid(True, alpha=0.3)

plt.suptitle('Step vs Sigmoid — Hard vs Soft Activation', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

</div>

### How the Algorithm Works — Step by Step

<div style="background: #0d1f2d; border: 1px solid #1e4d6b; border-radius: 10px; padding: 20px 26px; margin: 16px 0;">
  <div style="display: flex; flex-direction: column; gap: 10px; font-size: 13px;">
    <div style="display: flex; gap: 12px; align-items: flex-start;">
      <span style="background: #00d4ff; color: #0f2027; font-weight: 700; font-size: 11px; padding: 2px 8px; border-radius: 10px; white-space: nowrap;">Step 1</span>
      <span style="color: #7ecfdf;">Prepend a column of 1s to X — this becomes the bias term (intercept)</span>
    </div>
    <div style="display: flex; gap: 12px; align-items: flex-start;">
      <span style="background: #00d4ff; color: #0f2027; font-weight: 700; font-size: 11px; padding: 2px 8px; border-radius: 10px; white-space: nowrap;">Step 2</span>
      <span style="color: #7ecfdf;">Initialise all weights to 1 (or random)</span>
    </div>
    <div style="display: flex; gap: 12px; align-items: flex-start;">
      <span style="background: #00d4ff; color: #0f2027; font-weight: 700; font-size: 11px; padding: 2px 8px; border-radius: 10px; white-space: nowrap;">Step 3</span>
      <span style="color: #7ecfdf;">Pick a random sample j — compute prediction ŷ = step(w · xⱼ)</span>
    </div>
    <div style="display: flex; gap: 12px; align-items: flex-start;">
      <span style="background: #00d4ff; color: #0f2027; font-weight: 700; font-size: 11px; padding: 2px 8px; border-radius: 10px; white-space: nowrap;">Step 4</span>
      <span style="color: #7ecfdf;">Update weights: w ← w + η · (yⱼ − ŷ) · xⱼ. If prediction is correct, error=0 and weights don't change.</span>
    </div>
    <div style="display: flex; gap: 12px; align-items: flex-start;">
      <span style="background: #00d4ff; color: #0f2027; font-weight: 700; font-size: 11px; padding: 2px 8px; border-radius: 10px; white-space: nowrap;">Step 5</span>
      <span style="color: #7ecfdf;">Repeat for N epochs — the boundary rotates and shifts until classes are separated</span>
    </div>
  </div>
</div>

```python
def perceptron(X, y, lr=0.01, epochs=1000):
    """
    Perceptron classifier trained via online (stochastic) updates.

    Parameters
    ----------
    X      : feature matrix (n_samples, n_features)
    y      : binary labels {0, 1}
    lr     : learning rate
    epochs : number of random sample updates

    Returns
    -------
    intercept_, coef_ : bias scalar and weight vector
    """
    # Prepend bias column → weights[0] = intercept
    X_b     = np.insert(X, 0, 1, axis=1)          # (n, p+1)
    weights = np.ones(X_b.shape[1])               # initialise all to 1

    for _ in range(epochs):
        j     = np.random.randint(0, X_b.shape[0])  # random sample
        y_hat = step(np.dot(X_b[j], weights))        # predict
        weights += lr * (y[j] - y_hat) * X_b[j]     # update

    return weights[0], weights[1:]                   # intercept, coef
```

```python
intercept_, coef_ = perceptron(X, y, lr=0.01, epochs=1000)

print(f"Learned coefficients : {coef_}")
print(f"Learned intercept    : {intercept_:.4f}")
```

<a id='boundary'></a>
<div style="border-bottom: 2px solid #00d4ff; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #00d4ff; color: #0f2027; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 7</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">Plotting the Decision Boundary</h2>
</div>

The decision boundary is the line where `w · x + b = 0`. In 2D with features x₁ and x₂:

<div style="background: #0d1f2d; border: 1px solid #1e4d6b; border-radius: 10px; padding: 16px 24px; margin: 14px 0; text-align: center;">
  <div style="color: #7ecfdf; font-size: 12px; letter-spacing: 1px; margin-bottom: 8px;">Solving for x₂ (the y-axis)</div>
  <div style="color: #00d4ff; font-size: 16px; font-family: Georgia, serif;">
    w₁x₁ + w₂x₂ + b = 0 &nbsp;→&nbsp; x₂ = −(w₁/w₂)·x₁ − (b/w₂)
  </div>
  <div style="color: #7ecfdf; font-size: 12px; margin-top: 8px;">So: slope m = −w₁/w₂ &nbsp;&nbsp;|&nbsp;&nbsp; intercept b_line = −b/w₂</div>
</div>

```python
# Convert weight vector to slope-intercept form
m_perc = -(coef_[0] / coef_[1])
b_perc = -(intercept_ / coef_[1])

print(f"Decision boundary: x₂ = {m_perc:.3f}·x₁ + {b_perc:.3f}")

x_line = np.linspace(-3, 3, 100)
y_line = m_perc * x_line + b_perc

plt.figure(figsize=(9, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='winter', s=100, edgecolors='white', linewidth=0.5, zorder=3)
plt.plot(x_line, y_line, color='red', linewidth=2.5, label='Perceptron boundary', zorder=2)
plt.xlabel('Feature 1', fontsize=12)
plt.ylabel('Feature 2', fontsize=12)
plt.title('Perceptron — Learned Decision Boundary', fontsize=13, fontweight='bold')
plt.legend(fontsize=11)
plt.ylim(-3, 2)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

One of the most intuitive ways to understand the Perceptron is to **watch the decision boundary rotate and shift** during training. We record the boundary line at every epoch and animate it with `FuncAnimation`.

<div style="background: #0d1f2d; border: 1px solid #1e4d6b; border-radius: 8px; padding: 12px 18px; margin: 14px 0; font-size: 13px; color: #7ecfdf;">
  <strong style="color: #e6f4f8;">⚙️ Note:</strong> We use a higher learning rate (<code>lr=0.1</code>) and fewer epochs (<code>200</code>) so the boundary movement is visible and dramatic in the animation. Lower lr produces more stable but slower-to-visualise convergence.
</div>

```python
def perceptron_animated(X, y, lr=0.1, epochs=200):
    """
    Same as perceptron() but records the decision boundary at every epoch
    for animation purposes.
    """
    slopes, intercepts = [], []

    X_b     = np.insert(X, 0, 1, axis=1)
    weights = np.ones(X_b.shape[1])

    for _ in range(epochs):
        j     = np.random.randint(0, X_b.shape[0])
        y_hat = step(np.dot(X_b[j], weights))
        weights += lr * (y[j] - y_hat) * X_b[j]

        # Record slope and intercept of current boundary
        if weights[2] != 0:                               # avoid division by zero
            slopes.append(-(weights[1] / weights[2]))
            intercepts.append(-(weights[0] / weights[2]))
        else:
            slopes.append(slopes[-1] if slopes else 0)
            intercepts.append(intercepts[-1] if intercepts else 0)

    return slopes, intercepts
```

```python
m_hist, b_hist = perceptron_animated(X, y, lr=0.1, epochs=200)

x_anim = np.arange(-3, 3, 0.1)

fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(X[:, 0], X[:, 1], c=y, cmap='winter', s=100, edgecolors='white', linewidth=0.5, zorder=3)
line, = ax.plot(x_anim, x_anim * m_hist[0] + b_hist[0], 'r-', linewidth=2.5, zorder=2)
epoch_label = ax.set_xlabel('Epoch 1', fontsize=11)
ax.set_ylabel('Feature 2', fontsize=11)
ax.set_title('Perceptron — Decision Boundary Animation', fontsize=12, fontweight='bold')
ax.set_ylim(-3, 3)
ax.grid(True, alpha=0.3)

def update(i):
    line.set_ydata(x_anim * m_hist[i] + b_hist[i])
    ax.set_xlabel(f'Epoch {i + 1}', fontsize=11)

anim = FuncAnimation(fig, update, frames=200, interval=80, repeat=True)
plt.tight_layout()
plt.show()

# Save as gif (optional — comment out if not needed)
# anim.save('perceptron_learning.gif', writer='pillow', fps=15)
```

  <span style="background: #00d4ff; color: #0f2027; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 9</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">Logistic Regression — sklearn</h2>
</div>

Now we fit sklearn's `LogisticRegression` on the same data. Unlike the Perceptron, Logistic Regression minimises **binary cross-entropy (log loss)** using gradient descent — giving it stable, probabilistic outputs and guaranteed convergence.

```python
lor = LogisticRegression()
lor.fit(X, y)

print(f"Logistic Regression coefficients : {lor.coef_[0]}")
print(f"Logistic Regression intercept    : {lor.intercept_[0]:.4f}")
print(f"Training accuracy                : {accuracy_score(y, lor.predict(X)):.4f}")
```

```python
# Decision boundary from Logistic Regression
m_lor = -(lor.coef_[0][0] / lor.coef_[0][1])
b_lor = -(lor.intercept_[0] / lor.coef_[0][1])

print(f"Logistic boundary: x₂ = {m_lor:.3f}·x₁ + {b_lor:.3f}")
```

<a id='compare'></a>
<div style="border-bottom: 2px solid #00d4ff; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #00d4ff; color: #0f2027; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 10</span>
  <h2 style="color: #e6f4f8; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">Perceptron vs Logistic Regression — Comparison</h2>
</div>

The ultimate test: plot both decision boundaries on the same scatter plot and compare their accuracy and predictions.

```python
x_plot = np.linspace(-3, 3, 200)

# Perceptron boundary
y_perc = m_perc * x_plot + b_perc

# Logistic Regression boundary
y_lor  = m_lor  * x_plot + b_lor

plt.figure(figsize=(10, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='winter', s=100, edgecolors='white', linewidth=0.5, zorder=3, label='Data')
plt.plot(x_plot, y_perc, color='red',   linewidth=2.5, label='Perceptron boundary',           linestyle='--')
plt.plot(x_plot, y_lor,  color='black', linewidth=2.5, label='Logistic Regression boundary')
plt.xlabel('Feature 1', fontsize=12)
plt.ylabel('Feature 2', fontsize=12)
plt.title('Perceptron vs Logistic Regression — Decision Boundaries', fontsize=13, fontweight='bold')
plt.legend(fontsize=11)
plt.ylim(-3, 2)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ── Accuracy comparison ────────────────────────────────────────────────────
# Perceptron predictions using final learned weights
X_b_full = np.insert(X, 0, 1, axis=1)
w_full   = np.insert(coef_, 0, intercept_)
y_pred_perc = np.array([step(np.dot(X_b_full[i], w_full)) for i in range(len(X))])

acc_perc = accuracy_score(y, y_pred_perc)
acc_lor  = accuracy_score(y, lor.predict(X))

print(f"Perceptron         accuracy : {acc_perc:.4f}")
print(f"Logistic Regression accuracy : {acc_lor:.4f}")
print()
print("Logistic Regression Classification Report:")
print(classification_report(y, lor.predict(X)))
```
