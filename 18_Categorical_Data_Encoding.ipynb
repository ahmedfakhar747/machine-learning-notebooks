## 💡 Before We Begin...

Before encoding categorical features, it's important to understand the different types of categorical data. The type of data determines which encoding technique should be used.

There are **two main types of categorical data**:

### 📊 Ordinal Data

Ordinal data has a **meaningful order or ranking** between categories.

**Examples:**
- Poor → Average → Good → Excellent
- School → Diploma → Bachelor → Master → PhD

✅ **Recommended Encoding:** **Ordinal Encoding**



### 🏷️ Nominal Data

Nominal data has **no natural order**. The categories are simply labels.

**Examples:**
- Male / Female
- Red / Blue / Green
- Karachi / Lahore / Islamabad

✅ **Recommended Encoding:** **Label Encoding** *(for binary categories like Male/Female)* or **One-Hot Encoding** *(for multiple categories like cities or colors).*



### 🔖 What is Label Encoding?

**Label Encoding** assigns a unique numerical label to each category.

**Example:**

| Gender | Encoded Value |
|---------|--------------:|
| Male | 1 |
| Female | 0 |

> **Note:** The assigned numbers do **not** represent any order or importance. They are simply unique identifiers.



Choosing the correct encoding technique is essential because using the wrong method can introduce unintended relationships between categories and negatively affect model performance.

# 📊 Step 1 — Ordinal Encoding

In this step, we'll apply **Ordinal Encoding** to categorical features that have a **natural order or ranking**.

Unlike nominal categories, ordinal categories represent a progression (e.g., *Poor → Average → Good → Excellent*), so preserving this order is important. **Ordinal Encoding** converts these categories into numerical values while maintaining their relative ranking.

In our dataset, the following features are ordinal:
- ⭐ **Review:** Poor → Average → Good → Excellent
- 🎓 **Education:** School → Diploma → Bachelor → Master → PhD

After encoding, Machine Learning models can understand the order of these categories and use them effectively during training.

## 1.1 Seperating Features and Target
We'll first separate the **ordinal features** from the **target variable** before applying the encoding.

```python
X = df[["review", "education"]]
y = df.purchased
```

## 1.2 Splitting the Dataset

Next, we'll split the dataset into **training** and **testing** sets.

The encoder will be fitted **only on the training data** to prevent **data leakage**, and the learned mappings will then be applied to both the training and testing sets.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
```

## 1.3 Applying Ordinal Encoding

Now we'll use **OrdinalEncoder** to convert the ordinal categories into numerical values.

Since our categories have a predefined order, we'll explicitly specify that order using the `categories` parameter.


```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(
    categories=[
        ["Poor", "Average", "Good", "Excellent"],
        ["School", "High School", "Diploma", "Bachelor", "Master", "PhD"]
    ]
)

oe.fit(X_train)

X_train = oe.transform(X_train)
```

## 1.4 Encoded Features

After applying Ordinal Encoding, each category is replaced with its corresponding numerical value while preserving its order.

For example:

| Original | Encoded |
|----------|--------:|
| Poor | 0 |
| Average | 1 |
| Good | 2 |
| Excellent | 3 |

| Original | Encoded |
|----------|--------:|
| School | 0 |
| High School | 1 |
| Diploma | 2 |
| Bachelor | 3 |
| Master | 4 |
| PhD | 5 |


```python
pd.DataFrame(X_train, columns=[["review", "education"]])
```

## 1.5 Learned Categories

You can also inspect the category order learned by the encoder.


```python
oe.categories_
```

# 🎯 Step 2 — Label Encoder (Encoding the Target Variable)

Our target variable (`Purchased`) contains only two categories: **Yes** and **No**.

For binary target variables, **Label Encoding** is a simple and effective choice because it converts each class into a unique numerical label.


```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
le.fit(y_train)
```

## 2.1 Learned Classes

Let's see how the target classes are mapped.


```python
le.classes_
```

The classes are encoded as:

- **No → 0**
- **Yes → 1**

This numerical representation allows Machine Learning algorithms to use the target variable during training.
