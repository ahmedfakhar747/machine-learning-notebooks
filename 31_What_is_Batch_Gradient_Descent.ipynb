# 🧠 What is Batch Gradient Descent?

**Batch Gradient Descent** calculates the gradient using the **entire training dataset** before updating the model parameters.

The process is:

```text
Initialize weights and bias
          ↓
Calculate predictions
          ↓
Calculate error
          ↓
Calculate gradients
          ↓
Update weights and bias
          ↓
Repeat for several epochs
```

Unlike **Stochastic Gradient Descent**, which updates parameters using one sample at a time, Batch Gradient Descent uses **all training samples for every update**.

# 5. Implement Batch Gradient Descent from Scratch

Now let's build our own Linear Regression model using **Batch Gradient Descent**.



````python
class GDRegressor:

    def __init__(self, learning_rate=0.01, epochs=100):
        self.coef_ = None
        self.intercept_ = None

        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X_train, y_train):

        # Initialize parameters
        self.coef_ = np.ones(X_train.shape[1])
        self.intercept_ = 0

        for i in range(self.epochs):

            # Predictions
            y_hat = np.dot(X_train, self.coef_) + self.intercept_

            # Gradients
            coef_der = -2 * np.mean(
                X_train * (y_train.reshape(-1, 1) - y_hat.reshape(-1, 1)),
                axis=0
            )

            intercept_der = -2 * np.mean(y_train - y_hat)

            # Update parameters
            self.coef_ = self.coef_ - (self.lr * coef_der)
            self.intercept_ = self.intercept_ - (self.lr * intercept_der)

        print(self.coef_, self.intercept_)

    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
````

---

# 6. Understanding the Gradient Calculation

For Linear Regression, the Mean Squared Error is:

### $MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2$

The gradient with respect to each weight is:

### $\frac{\partial MSE}{\partial w_j} = -\frac{2}{n}\sum_{i=1}^{n}x_{ij}(y_i-\hat{y}_i)$

And for the intercept:

### $\frac{\partial MSE}{\partial b} = -\frac{2}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)$

The parameters are then updated using:

### $w_j = w_j - \alpha \frac{\partial MSE}{\partial w_j}$

### $b = b - \alpha \frac{\partial MSE}{\partial b}$

where:

* $( \alpha )$ = learning rate
* $( w_j )$ = model weight
* $( b )$ = intercept
* $( n )$ = number of training samples

Because we calculate these gradients using **the entire training dataset**, this is **Batch Gradient Descent**.

# 7. Train Our Custom Model

Let's now train our implementation.



````python
gd = GDRegressor(
    learning_rate=0.9, epochs=200
)

gd.fit(X_train, y_train)
````

---

# 9. Make Predictions



````python
gd_y_pred = gd.predict(X_test)
````

# 10. Evaluate Our Model

We'll use the same **R² score** used for the Scikit-learn model.




````python
gd_r2 = r2_score(y_test, gd_y_pred)

print("Batch Gradient Descent R² Score:", gd_r2)
````

# ⚠️ Important Note About the Learning Rate

The learning rate controls how large each parameter update is.

```python
learning_rate = 0.9
```

works reasonably well here because the Diabetes dataset features are already scaled.

For datasets where features have very different magnitudes, a large learning rate can cause Gradient Descent to **diverge** or behave unstably.

That's why **feature scaling is usually important when using Gradient Descent**.
