# 🗳️ 👥 ML 47 — Voting Ensemble | Intuition & Code

## Importing Libraries

```python
import numpy as np
import pandas as pd

from sklearn.datasets import make_classification
from sklearn.datasets import fetch_california_housing

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.model_selection import cross_val_score
```

## Reading Data

```python
df = pd.read_csv(
    '/kaggle/input/datasets/wasiqaliyasir/breast-cancer-dataset/Breast_cancer_dataset.csv',
    usecols=lambda column: column not in ["Unnamed: 32"]
)
df.head()
```

## Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df.diagnosis = encoder.fit_transform(df.diagnosis)

df.head()
```

## Seperating Features & Target

```python
X = df.drop(columns=['diagnosis'])
y = df.diagnosis
```

# Voting Ensemble | Classification

## Initializing Algorithms

```python
clf1 = LogisticRegression(max_iter=250)
clf2 = RandomForestClassifier()
clf3 = GaussianNB()
```

## Setting Estimators

```python
estimators = [('LogisticRegression', clf1), ('RandomForestClassifier', clf2), ('GaussianNB', clf3)]
```

```python
for estimator in estimators:
    x = cross_val_score(estimator[1], X, y, cv=10, scoring='accuracy')
    print(estimator[0], ': ', np.round(np.mean(x), 2))
```

```python
from sklearn.ensemble import VotingClassifier
```

## Hard Voting

```python
vc = VotingClassifier(estimators=estimators)

x = cross_val_score(vc, X, y, cv=10, scoring='accuracy')
print(np.round(np.mean(x), 2))
```

## Soft Voting

```python
vc = VotingClassifier(estimators=estimators, voting='soft')

x = cross_val_score(vc, X, y, cv=10, scoring='accuracy')
print(np.round(np.mean(x), 2))
```

## Weighted Voting

```python
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vc = VotingClassifier(estimators=estimators, voting='soft', weights=[i, j, k])
            x = cross_val_score(vc, X, y, cv=10, scoring='accuracy')
            print(f"for i={i}, j={j}, k={k}", np.round(np.mean(x), 2))
```

## Using Same Algorithm

```python
X, y = make_classification(n_samples=1000, n_features=20, n_informative=15, n_redundant=5, random_state=2)
```

# Voting Ensemble | Regression

# Reading Data

```python
X, y = fetch_california_housing(return_X_y=True)
```

```python
X.shape
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
rf = RandomForestRegressor(n_estimators=10)
```

```python
estimators = [('LinearRegression', lr), ('DecisionTreeRegressor', dt), ('RandomForestRegressor', rf)]
```

```python
for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
```

```python
from sklearn.ensemble import VotingRegressor
```

```python
vr = VotingRegressor(estimators)

scores = cross_val_score(vr, X, y, scoring='r2', cv=10)
print("Voting Regressor", np.round(np.mean(scores), 2))
```

```python
dt1 = DecisionTreeRegressor(max_depth=1)
dt2 = DecisionTreeRegressor(max_depth=3)
dt3 = DecisionTreeRegressor(max_depth=5)
dt4 = DecisionTreeRegressor(max_depth=7)
dt5 = DecisionTreeRegressor(max_depth=None)
```

```python
estimators = [('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5)]
```

```python
for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
```

```python
vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
```
