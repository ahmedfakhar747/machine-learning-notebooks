<a id='intuition'></a>
<div style="border-bottom: 2px solid #4f46e5; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #4f46e5; color: #fff; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 1</span>
  <h2 style="color: #e0e7ff; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">💡 Intuition &amp; Theoretical Concepts</h2>
</div>

Voting Ensembles combine individual estimators to generate a final prediction that outperforms any single constituent model.

<div style="display: grid; grid-template-columns: 1fr 1fr 1fr; gap: 12px; margin: 20px 0;">
  <div style="background: #090d16; border: 1px solid #1e1b4b; border-radius: 8px; padding: 16px;">
    <div style="color: #818cf8; font-weight: 700; font-size: 14px; margin-bottom: 6px;">✋ Hard Voting</div>
    <div style="color: #c7d2fe; font-size: 13px; line-height: 1.6;">Aggregates discrete class label predictions from each model and chooses the class with the majority of votes: $\hat{y} = \text{mode}\{C_1(x), C_2(x), \dots, C_M(x)\}$.</div>
  </div>
  <div style="background: #090d16; border: 1px solid #1e1b4b; border-radius: 8px; padding: 16px;">
    <div style="color: #818cf8; font-weight: 700; font-size: 14px; margin-bottom: 6px;">🌊 Soft Voting</div>
    <div style="color: #c7d2fe; font-size: 13px; line-height: 1.6;">Averages predicted class probabilities across all estimators and assigns the class with the highest average probability: $\hat{y} = \arg\max_c \frac{1}{M} \sum_{m=1}^M P_m(y=c|x)$.</div>
  </div>
  <div style="background: #090d16; border: 1px solid #1e1b4b; border-radius: 8px; padding: 16px;">
    <div style="color: #818cf8; font-weight: 700; font-size: 14px; margin-bottom: 6px;">📈 Voting Regressor</div>
    <div style="color: #c7d2fe; font-size: 13px; line-height: 1.6;">Computes the unweighted or weighted mean prediction across all continuous regression models: $\hat{y} = \frac{1}{M} \sum_{m=1}^M R_m(x)$.</div>
  </div>
</div>

```python
import numpy as np
import pandas as pd

# Datasets
from sklearn.datasets import make_classification
from sklearn.datasets import fetch_california_housing

# Classification Models
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.ensemble import VotingClassifier

# Regression Models
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import VotingRegressor

# Evaluation & Preprocessing
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import LabelEncoder
```

---

<a id='data-prep'></a>
<div style="border-bottom: 2px solid #4f46e5; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #4f46e5; color: #fff; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 3</span>
  <h2 style="color: #e0e7ff; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">📂 Data Preprocessing (Breast Cancer Dataset)</h2>
</div>

Load the Wisconsin Breast Cancer dataset and drop non-informative columns like `Unnamed: 32`.

```python
df = pd.read_csv(
    '/kaggle/input/datasets/wasiqaliyasir/breast-cancer-dataset/Breast_cancer_dataset.csv',
    usecols=lambda column: column not in ["Unnamed: 32"]
)
df.head()
```

Encode the categorical binary target column (`diagnosis`: Malignant `'M'` vs. Benign `'B'`) into numeric format (`1` / `0`).

```python
encoder = LabelEncoder()
df.diagnosis = encoder.fit_transform(df.diagnosis)
df.head()
```

Separate predictors $X$ from the target labels $y$.

```python
X = df.drop(columns=['diagnosis'])
y = df.diagnosis
```

---

<a id='hard-soft-clf'></a>
<div style="border-bottom: 2px solid #4f46e5; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #4f46e5; color: #fff; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 4</span>
  <h2 style="color: #e0e7ff; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">🗳️ Voting Classifier (Heterogeneous Algorithms)</h2>
</div>

Heterogeneous ensembling combines distinctly different algorithmic families (e.g., linear models, tree-based models, and probabilistic classifiers) to maximize estimator diversity.

```python
# 1. Initialize diverse base classifiers
clf1 = LogisticRegression(max_iter=250)
clf2 = RandomForestClassifier()
clf3 = GaussianNB()

# 2. Package into estimator tuples (name, model)
estimators = [
    ('LogisticRegression', clf1),
    ('RandomForestClassifier', clf2),
    ('GaussianNB', clf3)
]
```

Evaluate baseline performance for each individual model using 10-fold Cross-Validation:

```python
for estimator in estimators:
    scores = cross_val_score(estimator[1], X, y, cv=10, scoring='accuracy')
    print(f"{estimator[0]} : {np.round(np.mean(scores), 2)}")
```

### 1. Hard Voting Classifier
Aggregates majority votes across Logistic Regression, Random Forest, and Gaussian Naive Bayes.

```python
vc_hard = VotingClassifier(estimators=estimators, voting='hard')
scores_hard = cross_val_score(vc_hard, X, y, cv=10, scoring='accuracy')
print("Hard Voting Accuracy:", np.round(np.mean(scores_hard), 2))
```

### 2. Soft Voting Classifier
Averages predicted probabilities across estimators.

```python
vc_soft = VotingClassifier(estimators=estimators, voting='soft')
scores_soft = cross_val_score(vc_soft, X, y, cv=10, scoring='accuracy')
print("Soft Voting Accuracy:", np.round(np.mean(scores_soft), 2))
```

<a id='weighted-voting'></a>
<div style="border-bottom: 2px solid #4f46e5; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #4f46e5; color: #fff; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 5</span>
  <h2 style="color: #e0e7ff; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">⚖️ Weighted Voting Hyperparameter Optimization</h2>
</div>

Weighted soft voting assigns importance weights $[w_1, w_2, w_3]$ to individual base models based on their reliability:

$$\hat{P}(y=c) = \frac{\sum_{m=1}^M w_m P_m(y=c)}{\sum_{m=1}^M w_m}$$

Here, we grid-search optimal model weight combinations:

```python
for i in range(1, 4):
    for j in range(1, 4):
        for k in range(1, 4):
            vc_weighted = VotingClassifier(estimators=estimators, voting='soft', weights=[i, j, k])
            scores = cross_val_score(vc_weighted, X, y, cv=10, scoring='accuracy')
            print(f"Weights [LR={i}, RF={j}, GNB={k}] -> Accuracy: {np.round(np.mean(scores), 2)}")
```

<a id='homo-clf'></a>
<div style="border-bottom: 2px solid #4f46e5; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #4f46e5; color: #fff; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 6</span>
  <h2 style="color: #e0e7ff; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">🔄 Voting Classifier (Homogeneous Models — Polynomial Support Vector Machines)</h2>
</div>

Ensembles can also combine instances of the **same algorithm** configured with different hyperparameters (e.g., polynomial kernels of degrees 1 through 5).

Generate a synthetic classification dataset:

```python
X_synth, y_synth = make_classification(
    n_samples=1000, 
    n_features=20, 
    n_informative=15, 
    n_redundant=5, 
    random_state=2
)
```

Instantiate 5 SVM classifiers with varying polynomial degrees:

```python
svm1 = SVC(probability=True, kernel='poly', degree=1)
svm2 = SVC(probability=True, kernel='poly', degree=2)
svm3 = SVC(probability=True, kernel='poly', degree=3)
svm4 = SVC(probability=True, kernel='poly', degree=4)
svm5 = SVC(probability=True, kernel='poly', degree=5)

svm_estimators = [
    ('SVM1', svm1),
    ('SVM2', svm2),
    ('SVM3', svm3),
    ('SVM4', svm4),
    ('SVM5', svm5)
]
```

Evaluate standalone accuracy for each polynomial degree:

```python
for name, model in svm_estimators:
    scores = cross_val_score(model, X_synth, y_synth, cv=10, scoring='accuracy')
    print(f"{name} (Degree {name[-1]}): {np.round(np.mean(scores), 2)}")
```

Evaluate Soft Voting Ensemble across all 5 SVMs:

```python
vc_svm = VotingClassifier(estimators=svm_estimators, voting='soft')
scores_vc_svm = cross_val_score(vc_svm, X_synth, y_synth, cv=10, scoring='accuracy')
print("Homogeneous SVM Voting Classifier Accuracy:", np.round(np.mean(scores_vc_svm), 2))
```

<a id='reg-hetero'></a>
<div style="border-bottom: 2px solid #4f46e5; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #4f46e5; color: #fff; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 7</span>
  <h2 style="color: #e0e7ff; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">📊 Voting Regressor (Heterogeneous Algorithms)</h2>
</div>

For continuous target variables, `VotingRegressor` computes the average prediction across base regression models.

Load the California Housing dataset:

```python
X_reg, y_reg = fetch_california_housing(return_X_y=True)
print("Dataset dimensions:", X_reg.shape)
```

Define heterogeneous regressors (Linear Regression, Decision Tree, and Random Forest):

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
rf = RandomForestRegressor(n_estimators=10)

reg_estimators = [
    ('LinearRegression', lr),
    ('DecisionTreeRegressor', dt),
    ('RandomForestRegressor', rf)
]
```

Evaluate single model baseline performance using $R^2$ score:

```python
for name, model in reg_estimators:
    scores = cross_val_score(model, X_reg, y_reg, scoring='r2', cv=10)
    print(f"{name} R² score: {np.round(np.mean(scores), 2)}")
```

Evaluate `VotingRegressor` ensemble performance:

```python
vr = VotingRegressor(reg_estimators)
scores_vr = cross_val_score(vr, X_reg, y_reg, scoring='r2', cv=10)
print("Voting Regressor R² Score:", np.round(np.mean(scores_vr), 2))
```

<a id='reg-homo'></a>
<div style="border-bottom: 2px solid #4f46e5; padding-bottom: 8px; margin: 32px 0 20px 0;">
  <span style="background: #4f46e5; color: #fff; font-size: 11px; font-weight: 700; letter-spacing: 1.5px; padding: 3px 10px; border-radius: 4px; text-transform: uppercase;">Section 8</span>
  <h2 style="color: #e0e7ff; font-size: 22px; font-weight: 700; margin: 10px 0 0 0;">🌲 Voting Regressor (Homogeneous Models — Tree Depths)</h2>
</div>

Combine multiple Decision Trees with varying tree depth constraints (`max_depth` = 1, 3, 5, 7, Unconstrained) to balance high-bias and high-variance models.

```python
dt1 = DecisionTreeRegressor(max_depth=1)
dt2 = DecisionTreeRegressor(max_depth=3)
dt3 = DecisionTreeRegressor(max_depth=5)
dt4 = DecisionTreeRegressor(max_depth=7)
dt5 = DecisionTreeRegressor(max_depth=None)

dt_estimators = [
    ('dt1', dt1),
    ('dt2', dt2),
    ('dt3', dt3),
    ('dt4', dt4),
    ('dt5', dt5)
]
```

Evaluate baseline performance for each individual tree depth:

```python
for name, model in dt_estimators:
    scores = cross_val_score(model, X_reg, y_reg, scoring='r2', cv=10)
    print(f"{name} (max_depth={model.max_depth}): R² = {np.round(np.mean(scores), 2)}")
```

Evaluate `VotingRegressor` ensemble across all tree depths:

```python
vr_dt = VotingRegressor(dt_estimators)
scores_vr_dt = cross_val_score(vr_dt, X_reg, y_reg, scoring='r2', cv=10)
print("Homogeneous Decision Tree Voting Regressor R² Score:", np.round(np.mean(scores_vr_dt), 2))
```
