# 🗳️ 👥 ML 47 — Voting Ensemble | Intuition & Code

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(
    '/kaggle/input/datasets/wasiqaliyasir/breast-cancer-dataset/Breast_cancer_dataset.csv',
    usecols=lambda column: column not in ["Unnamed: 32"]
)
df.head()
```

# Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df.diagnosis = encoder.fit_transform(df.diagnosis)

df.head()
```

```python
X = df.drop(columns=['diagnosis'])
y = df.diagnosis
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.model_selection import cross_val_score
```

```python
clf1 = LogisticRegression(max_iter=1000)
clf2 = RandomForestClassifier()
clf3 = KNeighborsClassifier()
```

```python
estimators = [('LogisticRegression', clf1), ('RandomForestClassifier', clf2), ('KNeighborsClassifier', clf3)]
```

```python
for estimator in estimators:
    x = cross_val_score(estimator[1], X, y, cv=10, scoring='accuracy')
    print(estimator[0], np.round(np.mean(x), 2))
```
