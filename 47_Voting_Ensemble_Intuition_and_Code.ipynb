# 🗳️ 👥 ML 47 — Voting Ensemble | Intuition & Code

## Importing Libraries

```python
import numpy as np
import pandas as pd

from sklearn.datasets import make_classification

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.model_selection import cross_val_score
```

## Reading Data

```python
df = pd.read_csv(
    '/kaggle/input/datasets/wasiqaliyasir/breast-cancer-dataset/Breast_cancer_dataset.csv',
    usecols=lambda column: column not in ["Unnamed: 32"]
)
df.head()
```

## Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df.diagnosis = encoder.fit_transform(df.diagnosis)

df.head()
```

## Seperating Features & Target

```python
X = df.drop(columns=['diagnosis'])
y = df.diagnosis
```

# Voting Ensemble | Classification

## Initializing Algorithms

```python
clf1 = LogisticRegression(max_iter=250)
clf2 = RandomForestClassifier()
clf3 = GaussianNB()
```

## Setting Estimators

```python
estimators = [('LogisticRegression', clf1), ('RandomForestClassifier', clf2), ('GaussianNB', clf3)]
```

```python
for estimator in estimators:
    x = cross_val_score(estimator[1], X, y, cv=10, scoring='accuracy')
    print(estimator[0], ': ', np.round(np.mean(x), 2))
```

```python
from sklearn.ensemble import VotingClassifier
```

## Hard Voting

```python
vc = VotingClassifier(estimators=estimators)

x = cross_val_score(vc, X, y, cv=10, scoring='accuracy')
print(np.round(np.mean(x), 2))
```

## Soft Voting

```python
vc = VotingClassifier(estimators=estimators, voting='soft')

x = cross_val_score(vc, X, y, cv=10, scoring='accuracy')
print(np.round(np.mean(x), 2))
```

## Weighted Voting

```python
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vc = VotingClassifier(estimators=estimators, voting='soft', weights=[i, j, k])
            x = cross_val_score(vc, X, y, cv=10, scoring='accuracy')
            print(f"for i={i}, j={j}, k={k}", np.round(np.mean(x), 2))
```

## Using Same Algorithm

```python
X, y = make_classification(n_samples=1000, n_features=20, n_informative=15, n_redundant=5, random_state=2)
```

```python
svm1 = SVC(probability=True, kernel='poly', degree=1)
svm2 = SVC(probability=True, kernel='poly', degree=2)
svm3 = SVC(probability=True, kernel='poly', degree=3)
svm4 = SVC(probability=True, kernel='poly', degree=4)
svm5 = SVC(probability=True, kernel='poly', degree=5)
```

```python
estimators = [('SVM1', svm1), ('SVM2', svm2), ('SVM3', svm3), ('SVM4', svm4), ('SVM5', svm5)]
```

```python
for estimator in estimators:
    x = cross_val_score(estimator[1], X, y, cv=10, scoring='accuracy')
    print(estimator[0], np.round(np.mean(x), 2))
```

```python
vc1 = VotingClassifier(estimators=estimators, voting='soft')

x = cross_val_score(vc1, X, y, cv=10, scoring='accuracy')
print(np.round(np.mean(x), 2))
```
